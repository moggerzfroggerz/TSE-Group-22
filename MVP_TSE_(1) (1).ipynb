# **Chunking :**

```python
import os
import json
import uuid
from pathlib import Path
from typing import Dict, List

RAW_DATA_DIR = f"{base_path}/data/raw"
OUTPUT_DIR = f"{base_path}/data/processed"
OUTPUT_FILE = os.path.join(OUTPUT_DIR, "chunks.jsonl")

CHUNK_SIZE = 700
CHUNK_OVERLAP = 120


def ensure_dir(path: str) -> None:
    os.makedirs(path, exist_ok=True)


def read_text_file(path: str) -> str:
    with open(path, "r", encoding="utf-8") as f:
        return f.read()


def parse_txt_document(content: str, file_path: str) -> Dict:
    lines = [line.rstrip() for line in content.splitlines()]

    metadata = {
        "title": "",
        "source": "",
        "mode": "",
        "topic": "",
    }

    body_lines = []
    metadata_finished = False

    for line in lines:
        stripped = line.strip()

        if not metadata_finished:
            if stripped.startswith("Title:"):
                metadata["title"] = stripped.replace("Title:", "", 1).strip()
            elif stripped.startswith("Source:"):
                metadata["source"] = stripped.replace("Source:", "", 1).strip()
            elif stripped.startswith("Mode:"):
                metadata["mode"] = stripped.replace("Mode:", "", 1).strip().lower()
            elif stripped.startswith("Topic:"):
                metadata["topic"] = stripped.replace("Topic:", "", 1).strip()
            elif stripped == "":
                continue
            else:
                metadata_finished = True
                body_lines.append(line)
        else:
            body_lines.append(line)

    body = "\n".join(body_lines).strip()

    if not metadata["title"]:
        metadata["title"] = Path(file_path).stem.replace("_", " ").title()
    if not metadata["source"]:
        metadata["source"] = "UNKNOWN"
    if not metadata["mode"]:
        metadata["mode"] = "patient"
    if not metadata["topic"]:
        metadata["topic"] = metadata["title"]

    return {
        "doc_id": str(uuid.uuid4()),
        "file_path": file_path,
        "title": metadata["title"],
        "source": metadata["source"],
        "mode": metadata["mode"],
        "topic": metadata["topic"],
        "text": body,
    }


def load_documents(raw_dir: str) -> List[Dict]:
    documents = []

    for root, _, files in os.walk(raw_dir):
        for file_name in files:
            if not file_name.endswith(".txt"):
                continue

            full_path = os.path.join(root, file_name)
            content = read_text_file(full_path)
            doc = parse_txt_document(content, full_path)
            documents.append(doc)

    return documents


def smart_split_text(text: str, chunk_size: int, overlap: int) -> List[str]:
    text = text.strip()
    if not text:
        return []

    if len(text) <= chunk_size:
        return [text]

    chunks = []
    start = 0
    text_len = len(text)

    while start < text_len:
        end = min(start + chunk_size, text_len)
        candidate = text[start:end]

        if end < text_len:
            paragraph_break = candidate.rfind("\n\n")
            if paragraph_break > int(chunk_size * 0.5):
                end = start + paragraph_break
            else:
                newline_break = candidate.rfind("\n")
                if newline_break > int(chunk_size * 0.5):
                    end = start + newline_break
                else:
                    sentence_break = candidate.rfind(". ")
                    if sentence_break > int(chunk_size * 0.5):
                        end = start + sentence_break + 1

        chunk = text[start:end].strip()
        if chunk:
            chunks.append(chunk)

        if end >= text_len:
            break

        start = max(end - overlap, start + 1)

    return chunks


def chunk_documents(documents: List[Dict], chunk_size: int, overlap: int) -> List[Dict]:
    all_chunks = []

    for doc in documents:
        chunks = smart_split_text(doc["text"], chunk_size, overlap)

        for idx, chunk_text in enumerate(chunks):
            all_chunks.append({
                "chunk_id": str(uuid.uuid4()),
                "doc_id": doc["doc_id"],
                "chunk_index": idx,
                "title": doc["title"],
                "source": doc["source"],
                "mode": doc["mode"],
                "topic": doc["topic"],
                "file_path": doc["file_path"],
                "text": chunk_text
            })

    return all_chunks


def save_jsonl(rows: List[Dict], output_path: str) -> None:
    with open(output_path, "w", encoding="utf-8") as f:
        for row in rows:
            f.write(json.dumps(row, ensure_ascii=False) + "\n")


def print_summary(documents: List[Dict], chunks: List[Dict]) -> None:
    print("\n=== DOCUMENT SUMMARY ===")
    print(f"Documents loaded: {len(documents)}")

    for doc in documents:
        print(
            f"- {doc['title']} | Source={doc['source']} | Mode={doc['mode']} | "
            f"Topic={doc['topic']} | Chars={len(doc['text'])}"
        )

    print("\n=== CHUNK SUMMARY ===")
    print(f"Total chunks created: {len(chunks)}")

    by_doc = {}
    for chunk in chunks:
        by_doc.setdefault(chunk["title"], 0)
        by_doc[chunk["title"]] += 1

    for title, count in by_doc.items():
        print(f"- {title}: {count} chunks")


def preview_chunks(chunks: List[Dict], max_preview: int = 3) -> None:
    print("\n=== PREVIEW ===")
    for chunk in chunks[:max_preview]:
        preview = chunk["text"][:250].replace("\n", " ")
        print(f"\n[{chunk['title']} | chunk {chunk['chunk_index']}]")
        print(preview + ("..." if len(chunk["text"]) > 250 else ""))


ensure_dir(OUTPUT_DIR)

documents = load_documents(RAW_DATA_DIR)
if not documents:
    raise ValueError("No .txt files found in data/raw")

chunks = chunk_documents(documents, CHUNK_SIZE, CHUNK_OVERLAP)
save_jsonl(chunks, OUTPUT_FILE)

print_summary(documents, chunks)
preview_chunks(chunks)

print(f"\nSaved chunk file to: {OUTPUT_FILE}")
```

##**Generate embeddings :**

```python
import numpy as np
from sentence_transformers import SentenceTransformer

model_name = "sentence-transformers/all-MiniLM-L6-v2"
model = SentenceTransformer(model_name)

texts = [chunk["text"] for chunk in chunks]

embeddings = model.encode(
    texts,
    batch_size=32,
    show_progress_bar=True,
    normalize_embeddings=True
)

embeddings = np.array(embeddings, dtype="float32")

print("Embeddings shape:", embeddings.shape)
```

##**Building and saving of FAISS Index :**

```python
import faiss

embedding_dim = embeddings.shape[1]

# cosine similarity is used through inner product on normalised embeddings
index = faiss.IndexFlatIP(embedding_dim)
index.add(embeddings)

faiss.write_index(index, faiss_index_path)

print(f"FAISS index built and saved to: {faiss_index_path}")
print(f"Total vectors in index: {index.ntotal}")
```

##**Loading of FAISS Index and metadata retrieval :**

```python
import faiss
import json

# Reload to simulate actual use
index = faiss.read_index(faiss_index_path)

meta = []
with open(meta_path, "r", encoding="utf-8") as f:
    for line in f:
        line = line.strip()
        if line:
            meta.append(json.loads(line))

print("Reloaded FAISS index with", index.ntotal, "vectors")
print("Reloaded metadata rows:", len(meta))
```

##**Retrieval function :**

```python
def retrieve_chunks(query, mode="patient", top_k=3):
    """
    Retrieve top chunks matching the query.
    Mode can be 'patient' or 'professional'.
    """
    query_embedding = model.encode([query], normalize_embeddings=True)
    query_embedding = np.array(query_embedding, dtype="float32")

    # retrieve more than needed, then filter by mode
    raw_k = min(max(top_k * 4, 10), index.ntotal)
    scores, indices = index.search(query_embedding, raw_k)

    results = []
    seen_texts = set()

    for score, idx in zip(scores[0], indices[0]):
        if idx < 0:
            continue

        row = meta[idx]

        if row["mode"] != mode:
            continue

        # simple dedup
        text_key = row["text"][:120]
        if text_key in seen_texts:
            continue
        seen_texts.add(text_key)

        results.append({
            "title": row["title"],
            "source": row["source"],
            "mode": row["mode"],
            "topic": row["topic"],
            "text": row["text"],
            "score": float(score)
        })

        if len(results) >= top_k:
            break

    return results
```

##**Answer formatter for the MVP Demo :**

```python
def format_mvp_answer(query, mode="patient", top_k=2):
    results = retrieve_chunks(query, mode=mode, top_k=top_k)

    if not results:
        return "No relevant information found."

    answer_parts = []
    sources = []

    for result in results:
        answer_parts.append(result["text"])
        sources.append(f"{result['source']} - {result['title']}")

    answer = "\n\n".join(answer_parts)
    source_text = "\n".join(f"- {s}" for s in sorted(set(sources)))

    final_output = f"""Question:
{query}

Mode:
{mode}

Retrieved Answer:
{answer}

Sources:
{source_text}

Safety Note:
This information is for educational purposes only and does not replace professional medical advice.
"""
    return final_output
```

##**Full API code :**

```python
import json
from contextlib import asynccontextmanager
from typing import Literal, List

import faiss
import numpy as np
from fastapi import FastAPI, HTTPException
from pydantic import BaseModel, Field
from sentence_transformers import SentenceTransformer


# =========================
# CONFIG
# =========================
EMBEDDING_MODEL_NAME = "sentence-transformers/all-MiniLM-L6-v2"
DEFAULT_TOP_K = 3
MAX_TOP_K = 8

SAFETY_NOTE = (
    "This information is for educational purposes only and does not replace "
    "professional medical advice."
)

EMERGENCY_MESSAGE = (
    "If this may be a medical emergency, seek urgent medical help immediately "
    "or call emergency services."
)


# =========================
# REQUEST / RESPONSE MODELS
# =========================
ModeType = Literal["patient", "professional"]


class RetrieveRequest(BaseModel):
    query: str = Field(..., min_length=2, max_length=1000)
    mode: ModeType = "patient"
    top_k: int = Field(default=DEFAULT_TOP_K, ge=1, le=MAX_TOP_K)


class ChunkResult(BaseModel):
    chunk_id: str
    title: str
    source: str
    mode: str
    topic: str
    text: str
    score: float


class RetrieveResponse(BaseModel):
    query: str
    mode: ModeType
    results: List[ChunkResult]


class AnswerRequest(BaseModel):
    query: str = Field(..., min_length=2, max_length=1000)
    mode: ModeType = "patient"
    top_k: int = Field(default=DEFAULT_TOP_K, ge=1, le=MAX_TOP_K)


class AnswerResponse(BaseModel):
    query: str
    mode: ModeType
    answer: str
    sources: List[str]
    emergency_flag: bool
    safety_note: str


# =========================
# SERVICE
# =========================
class RetrievalService:
    def __init__(self, index_path: str, meta_path: str, model_name: str):
        self.index_path = index_path
        self.meta_path = meta_path
        self.model_name = model_name

        self.index = None
        self.meta = []
        self.model = None

    def load(self):
        self.index = faiss.read_index(self.index_path)

        self.meta = []
        with open(self.meta_path, "r", encoding="utf-8") as f:
            for line in f:
                line = line.strip()
                if line:
                    self.meta.append(json.loads(line))

        self.model = SentenceTransformer(self.model_name)

        if self.index.ntotal != len(self.meta):
            raise ValueError(
                f"Index/meta mismatch: index has {self.index.ntotal} rows "
                f"but metadata has {len(self.meta)} rows."
            )

    def ready(self):
        return self.index is not None and self.model is not None and len(self.meta) > 0

    def detect_emergency(self, query: str) -> bool:
        emergency_terms = [
            "chest pain",
            "can't breathe",
            "cannot breathe",
            "severe shortness of breath",
            "stroke",
            "face drooping",
            "slurred speech",
            "one sided weakness",
            "one-sided weakness",
            "asthma attack",
            "unconscious",
            "seizure",
        ]
        q = query.lower()
        return any(term in q for term in emergency_terms)

    def retrieve(self, query: str, mode: str, top_k: int):
        if not self.ready():
            raise RuntimeError("Retrieval service is not loaded.")

        query_embedding = self.model.encode([query], normalize_embeddings=True)
        query_embedding = np.asarray(query_embedding, dtype="float32")

        raw_k = min(max(top_k * 4, 10), self.index.ntotal)
        scores, indices = self.index.search(query_embedding, raw_k)

        results = []
        seen_texts = set()

        for score, idx in zip(scores[0], indices[0]):
            if idx < 0:
                continue

            row = self.meta[idx]

            if str(row.get("mode", "")).lower() != mode:
                continue

            text = str(row.get("text", "")).strip()
            if not text:
                continue

            dedup_key = text[:150]
            if dedup_key in seen_texts:
                continue
            seen_texts.add(dedup_key)

            results.append({
                "chunk_id": str(row.get("chunk_id", "")),
                "title": str(row.get("title", "")),
                "source": str(row.get("source", "")),
                "mode": str(row.get("mode", "")),
                "topic": str(row.get("topic", "")),
                "text": text,
                "score": float(score),
            })

            if len(results) >= top_k:
                break

        return results

    def answer(self, query: str, mode: str, top_k: int):
        results = self.retrieve(query, mode, top_k)

        if not results:
            return {
                "query": query,
                "mode": mode,
                "answer": "No relevant information was found in the current EduCare AI dataset.",
                "sources": [],
                "emergency_flag": self.detect_emergency(query),
                "safety_note": SAFETY_NOTE,
            }

        body = "\n\n".join(r["text"] for r in results)

        intro = (
            "Here is educational information based on trusted healthcare sources:\n\n"
            if mode == "patient"
            else "Here is guideline-oriented information retrieved from the professional dataset:\n\n"
        )

        emergency_flag = self.detect_emergency(query)
        emergency_prefix = (
            f"Urgent Safety Advice:\n{EMERGENCY_MESSAGE}\n\n"
            if emergency_flag else ""
        )

        sources = sorted({f"{r['source']} — {r['title']}" for r in results})

        return {
            "query": query,
            "mode": mode,
            "answer": f"{emergency_prefix}{intro}{body}",
            "sources": sources,
            "emergency_flag": emergency_flag,
            "safety_note": SAFETY_NOTE,
        }


service = RetrievalService(
    index_path=FAISS_INDEX_PATH,
    meta_path=META_PATH,
    model_name=EMBEDDING_MODEL_NAME,
)


@asynccontextmanager
async def lifespan(app: FastAPI):
    service.load()
    app.state.retrieval_service = service
    yield


app = FastAPI(
    title="EduCare AI API",
    version="1.0.0",
    lifespan=lifespan,
)


@app.get("/health")
def health():
    return {"status": "ok"}


@app.post("/retrieve", response_model=RetrieveResponse)
def retrieve(payload: RetrieveRequest):
    try:
        results = app.state.retrieval_service.retrieve(
            query=payload.query,
            mode=payload.mode,
            top_k=payload.top_k,
        )
        return {
            "query": payload.query,
            "mode": payload.mode,
            "results": results,
        }
    except Exception as exc:
        raise HTTPException(status_code=500, detail=str(exc))


@app.post("/answer", response_model=AnswerResponse)
def answer(payload: AnswerRequest):
    try:
        result = app.state.retrieval_service.answer(
            query=payload.query,
            mode=payload.mode,
            top_k=payload.top_k,
        )
        return result
    except Exception as exc:
        raise HTTPException(status_code=500, detail=str(exc))
```

##**API helper function :**

```python
import requests

def check_api_health():
    try:
        response = requests.get(f"{API_BASE_URL}/health", timeout=30)
        response.raise_for_status()
        data = response.json()
        if data.get("status") == "ok":
            return "Connected"
        return f"Unexpected response: {data}"
    except Exception as e:
        return f"Disconnected: {str(e)}"


def call_answer_api(query: str, mode: str, top_k: int):
    payload = {
        "query": query,
        "mode": mode,
        "top_k": top_k
    }

    response = requests.post(
        f"{API_BASE_URL}/answer",
        json=payload,
        timeout=60
    )
    response.raise_for_status()
    return response.json()
```

```python
def run_educare(query, mode, top_k):
    if not query or not query.strip():
        return (
            "Please enter a healthcare question.",
            "No sources retrieved.",
            "No"
        )

    try:
        result = call_answer_api(query=query.strip(), mode=mode, top_k=int(top_k))

        answer = result.get("answer", "No answer returned.")
        sources = result.get("sources", [])
        emergency_flag = result.get("emergency_flag", False)

        if sources:
            sources_text = "\n".join(f"- {s}" for s in sources)
        else:
            sources_text = "No sources retrieved."

        emergency_text = "Yes" if emergency_flag else "No"

        return answer, sources_text, emergency_text

    except Exception as e:
        return (
            f"API error: {str(e)}",
            "No sources retrieved.",
            "No"
        )
```
